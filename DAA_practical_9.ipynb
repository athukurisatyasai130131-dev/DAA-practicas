```python
import heapq

def prim(graph, start=0):
    visited = set()
    min_heap = [(0, start, -1)]
    mst = []
    total_cost = 0

    while min_heap:
        weight, u, parent = heapq.heappop(min_heap)

        if u in visited:
            continue

        visited.add(u)
        total_cost += weight

        if parent != -1:
            mst.append((parent, u, weight))

        for v, w in graph[u]:
            if v not in visited:
                heapq.heappush(min_heap, (w, v, u))

    return mst, total_cost


graph = {
    0: [(1, 2), (3, 6)],
    1: [(0, 2), (2, 3), (3, 8), (4, 5)],
    2: [(1, 3), (4, 7)],
    3: [(0, 6), (1, 8), (4, 9)],
    4: [(1, 5), (2, 7), (3, 9)]
}

mst, cost = prim(graph)

print("Minimum Spanning Tree:")
for u, v, w in mst:
    print(f"{u} - {v} : {w}")

print("Total cost:", cost)
```
